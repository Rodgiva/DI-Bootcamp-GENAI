## 🌟 Exercise 1 : Matrix Operations

1. Calculate the determinant.
2. Find the inverse of the matrix.

```python
import numpy as np
import pandas as pd

rng = np.random.default_rng(seed=43)

matrix = rng.integers(low=0, high=10, size=(3, 3))
determinant = np.linalg.det(matrix)
inverse = np.linalg.inv(matrix) if determinant != 0 else None

print(f'Matrice: \n{matrix}\n')
print(f'Determinant: \n{determinant}\n')
print(f'Inverse: \n{inverse}')
```

## 🌟 Exercise 6 : Basic Hypothesis Testing

```python
productivity_before = np.random.normal(loc=50, scale=10, size=30)
productivity_after = productivity_before + np.random.normal(loc=5, scale=3, size=30)

# Your task is to formulate a hypothesis regarding the training program's effectiveness 
# and test it using basic statistical functions in NumPy.
```

Your task is to formulate a hypothesis regarding the training program's effectiveness and test it using basic statistical functions in NumPy.

- Null Hypothesis H0: The training has no effect → the mean of the differences (after - before) is equal to 0.
- Alternative Hypothesis H1: The training improves productivity → the mean of the differences is greater than 0.

```python
from scipy import stats

np.random.seed(42)

productivity_before = np.random.normal(loc=50, scale=10, size=30)
productivity_after = productivity_before + np.random.normal(loc=5, scale=3, size=30)

t_stat, p_value = stats.ttest_rel(productivity_after, productivity_before)

alpha = 0.05
print(f"t-statistic: {t_stat:.4f}, p-value: {p_value:.4f}")

if p_value < alpha:
    print("Reject the null hypothesis")
else:
    print("Fail to reject the null hypothesis")
```

## 🌟 Exercise 10 : Basic Visualization

Use Matplotlib to visualize a simple dataset created with NumPy (e.g., a line graph of random numbers).

```python
rng2 = np.random.default_rng()
# seed=50
arr = rng2.integers(low=0, high=100, size=(20,3))
df = pd.DataFrame(arr, columns=['A', 'B', 'C'])
df.head(100)
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(12, 6))

df.plot(ax=ax, marker='o', linewidth=1.5)

ax.set_xlabel('X', fontsize=12, fontweight='bold')
ax.set_ylabel('Y', fontsize=12, fontweight='bold')
ax.set_title('The beautiful graph', fontsize=14, fontweight='bold')
ax.grid(True, linestyle='--', alpha=0.6)

plt.tight_layout()
plt.show()
```
