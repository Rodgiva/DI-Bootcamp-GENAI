## 🌟 Exercise 1: Understanding Data Visualization

Data visualization make easier to understand and interpret the datas, identifying patterns, trends, anomalies, etc.
Line graphs are for quantitative values over a continuous interval or time span.

## 🌟 Exercise 2: Creating a Line Plot for Temperature Variation

```python
import matplotlib.pyplot as plt

days = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"]
temp = [25,27,31,30,28,26,25]
plt.plot(days, temp)
plt.xlabel("Days")
plt.ylabel("Temperature")
plt.title("The best title you have ever seen")
plt.show()
```

## 🌟 Exercise 3: Visualizing Monthly Sales with a Bar Chart

```python
sales_val = [5000, 5500, 6200, 7000, 7500, 7200, 6800, 6500, 5200, 5700, 6300, 6000]
months = ["January", "February", "March", "April", "May", "June",
        "July", "August", "September", "October", "November", "December"]

plt.bar(months, sales_val)
plt.xlabel("Months")
plt.ylabel("Sales Amount ($)")
plt.title("a title to the bar chart")
plt.show()
```

## 🌟 Exercise 4: Visualizing the Distribution of CGPA

```python
import pandas as pd
import seaborn as sb

df = pd.read_csv('Student Mental health.csv')
cgpa = df['What is your CGPA?']

sb.histplot(cgpa, bins=30, color='green')
plt.xlabel('CGPA')
plt.ylabel('Density')
plt.title('nb of CGPA')
plt.show()
```

## 🌟 Exercise 5: Comparing Anxiety Levels Across Different Genders

```python
sb.barplot(x = 'Do you have Anxiety?', y = 'Choose your gender', data = df)
plt.xlabel('gender')
plt.ylabel('yes or no')
plt.show()
```

## 🌟 Exercise 6: Exploring the Relationship Between Age and Panic Attacks

```python
sb.scatterplot(y = 'Do you have Panic attack?', x = 'Age', data = df)
plt.xlabel('Age')
plt.ylabel('Panick attack?')
plt.show()
```
