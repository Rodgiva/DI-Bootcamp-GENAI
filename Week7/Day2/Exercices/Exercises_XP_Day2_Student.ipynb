2) Discuss: How did the shift to modern NLP impact scalability and efficiency?

- The shift to modern NLP significantly improved scalability by replacing manual feature engineering with representation learning, allowing models to automatically extract useful patterns from raw text at scale.
- The move from RNNs/LSTMs to Transformers was a major efficiency gain: self-attention processes all tokens in a sequence simultaneously rather than step-by-step, enabling full hardware parallelism.
- This parallelism made easier to train much larger models on much larger datasets within reasonable time frames.
- Fine tuning further boosted efficiency by letting a single pre-trained model be reused across many downstream tasks, instead of training a new model from scratch for each one.


## 🌟 Exercise 2: LLM Architecture and Application Scenarios
For each, describe (a) core architectural differences, (b) a real application, (c) why it fits.

### BERT
- Architecture: Bidirectional Encoder, Pretrained
- Application: Sentiment Analysis, Search engine, NES/POS   
- Why: BERT is bidirectional attention gives it a richer contextual representation of each token than a model that only reads left-to-right, making it well-suited for token-level classification tasks.

### GPT
- Architecture: Decoder-only, autoregressive causal LM
- Application: chatbot, text/code generation
- Why: Chat and generation tasks require producing coherent text one token at a time, building on everything generated so far. GPT causal architecture matches this generation process naturally, since it was trained to predict what comes next given only past context, what is needed at inference time.

### T5
- TODO: Encoder-decoder, text-to-text framework
- Application: summarization, translation
- Why: These tasks involve transforming one full sequence into a different full sequence (a long document into a short summary, or a sentence in one language into another). The encoder builds a complete representation of the input, and the decoder generates a new, independent output sequence conditioned on it, something a decoder-only or encoder-only model isn’t naturally structured to do.


## 🌟 Exercise 3: Benefits and Ethics of Pre-training
- Benefits (explain each in your words):
  1) Pretraining on vast, diverse text exposes the model to many language patterns, topics, and styles, so it learns more general linguistic knowledge rather than overfitting to one narrow task, making it perform better on unseen data.
  2) Less labeled data: The model already understands language structure from pre-training: fine-tuning only requires a small labeled dataset to adapt it to a specific task, instead of needing massive labeled data from scratch.
  3) Faster fine-tuning: Same: fine-tuning only needs a few additional training steps/epochs to adapt the model, saving time and compute compared to training from scratch.
  4) Transfer learning: Knowledge learned during pre-training on general language understanding transfers to many different downstream tasks, so one pre-trained model can be reused as the foundation for classification, NER, QA, summarization, etc.
  5) Robustness: Having been exposed to diverse text during pre-training, the model is less sensitive to noise, typos, or unusual phrasing in new data, making it more robust than a model trained only on a narrow, specific dataset.

- Ethical concerns: Pre-trained models can inherit and amplify societal biases present in web-scale training data, generate convincing but false information (hallucinations), be misused, and risk leaking private information that was memorized from the training corpus.

- Mitigations: 
* Data curation: cleaning training data of biased/toxic/private content
* Differential privacy: preventing memorization of individual data points
* Safety filters on outputs, RLHF to align model behavior with human values, and regular independent audits to detect bias or harmful behavior.

### Self-Attention & Multi-Head Attention
- TODO: Describe the Q/K/V flow, softmax weighting, and value mixing: Each token is projected into three vectors, Query, Key, and Value. The attention score between two tokens is computed as the dot product of one token’s Query with another’s Key, scaled by the square root of the key dimension. These scores are passed through softmax to produce attention weights that sum to 1. Each token’s final representation is then a weighted sum of all tokens’ Value vectors, using these softmax weights — so tokens that are more relevant (higher Q·K score) contribute more to the mix.

### (Optional) Quick Generation Demo Across Temperatures
Note: Requires `transformers` and model download; skip if offline.


```python
from transformers import AutoTokenizer, AutoModelForCausalLM, pipeline
import torch

model_name = "gpt2"  # TODO: choose a causal LM
prompt = "Artificial intelligence will"  # TODO: write your own prompt

pipe = pipeline(
    "text-generation",
    model=AutoModelForCausalLM.from_pretrained(model_name),
    tokenizer=AutoTokenizer.from_pretrained(model_name),
    device=0 if torch.cuda.is_available() else -1,
)

for T in [0.2, 1.0, 1.5]:
    out = pipe(prompt, max_new_tokens=40, temperature=T, do_sample=True)
    print("--- Temperature:", T)
    print(out[0]["generated_text"])  # Inspect how style changes with T
```
