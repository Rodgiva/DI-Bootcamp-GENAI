## 🌟 Exercise 2: Descriptive Statistics

```python
data = [12, 15, 13, 12, 18, 20, 22, 21]
mean_data = stats.tmean(data)
median_data = stats.scoreatpercentile(data, 50)
variance_data = stats.tvar(data)
deviation_data = stats.tstd(data)

print(f"mean: {mean_data}")
print(f"median: {median_data}")
print(f"variance: {variance_data}")
print(f"deviatio: {deviation_data}")
```
