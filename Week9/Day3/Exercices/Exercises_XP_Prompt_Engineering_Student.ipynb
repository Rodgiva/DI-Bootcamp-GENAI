## Exercise 2: Multi-Part Prompt for Quiz Generation
**Scenario**: Create an educational prompt for 7th graders about volcanic eruptions.

### Tasks
- Ask for: 2-bullet summary + 3 MCQs (1 correct, 2 distractors).
- Require friendly tone, simple vocab (11?13 y/o), and Google Slides-ready output.
- Explain 3 ways your prompt improves on: `Make a quiz for kids about this article.`

```python
quiz_prompt = """You are a middle-school science teacher creating slide content for 7th graders (11-13 years old).

Source material: the article on volcanic eruptions provided below. Use only the information it contains.

Produce two parts:

PART 1 - Summary
- Exactly 2 bullet points.
- Each bullet is one sentence, maximum 20 words.
- Explain what causes a volcanic eruption and what happens during one.

PART 2 - Quiz
- Exactly 3 multiple-choice questions, numbered 1 to 3.
- Each question has 3 options labelled A, B, C: exactly 1 correct answer and 2 plausible distractors.
- Distractors must be believable but clearly wrong according to the article, not jokes.
- After the 3 questions, add an "Answer Key" line in the form 1-A, 2-C, 3-B.

Style and tone:
- Friendly and encouraging, as if speaking directly to the student.
- Simple vocabulary for an 11-13 year old reading level; short sentences.
- Explain any technical term (magma, tectonic plate, lava) in a few words the first time it appears.

Output format (Google Slides ready):
- Plain text only, no markdown, no bold, no tables.
- One slide per block, separated by a line containing only "---".
- Slide 1 title: "Volcanic Eruptions: Key Ideas", then the 2 summary bullets.
- Slides 2 to 4: one question per slide, title "Question N", then the 3 options on their own lines.
- Slide 5 title: "Answer Key", then the answer line.
- Each line must be short enough to fit on a slide (max ~12 words per bullet).

Grounding rule: if the article does not contain the information needed for a question,
write "NOT IN SOURCE" instead of inventing a fact.

ARTICLE:
[PASTE ARTICLE TEXT HERE]"""
quiz_prompt
```

```python
improvements = [
    'Specific structure and quantity: it fixes exactly 2 summary bullets and 3 MCQs with 3 options, 1 correct and 2 distractors, plus an answer key, instead of leaving length and shape to chance.',
    'Defined audience and tone: "7th graders, 11-13 years old, friendly, simple vocabulary, define technical terms" replaces the vague word "kids", so the reading level is controlled rather than guessed.',
    'Explicit output format and grounding: plain text, one slide per block separated by ---, short lines for Google Slides, and a rule to answer "NOT IN SOURCE" rather than invent facts, which makes the result directly usable and reduces hallucination.',
]
improvements
```

## Exercise 4: Match Prompt to Purpose
Pick a style (Exploratory, Structured, Conversational, Functional) to build a customer-support chatbot prompt.

### Tasks
- Write the full prompt in the chosen style.
- Identify 2 features that prove it matches the style.
- Justify why this style fits better than the others.

```python
chosen_style = 'Functional'

prompt_template = """You are the first-line customer support assistant for FlowNest, a task and note
management SaaS. You handle incoming customer messages in a web chat widget.

Operating rules:
1. Read the customer message and classify it into exactly one category:
   BILLING, BUG, HOW_TO, ACCOUNT_ACCESS, FEATURE_REQUEST, OTHER.
2. Answer using only the FlowNest help-centre excerpts provided in CONTEXT below.
3. If CONTEXT does not cover the question, do not guess: set escalate to true and tell the
   customer a human agent will follow up within one business day.
4. Never ask for, repeat or store passwords, full card numbers or any other credential.
5. Never promise refunds, discounts, deadlines or roadmap items. Those always escalate.
6. Keep the reply under 120 words, plain language, polite and neutral. Address the customer as "you".
7. If information is missing to resolve the issue, ask at most 2 clarifying questions, numbered.

Respond with valid JSON only, no text outside the object, using exactly this schema:
{
  "category": "BILLING | BUG | HOW_TO | ACCOUNT_ACCESS | FEATURE_REQUEST | OTHER",
  "reply": "<message shown to the customer>",
  "clarifying_questions": ["<question>", "..."],
  "escalate": true | false,
  "escalation_reason": "<short reason, empty string if escalate is false>",
  "confidence": 0.0
}

CONTEXT:
[PASTE RELEVANT HELP-CENTRE ARTICLES HERE]

CUSTOMER MESSAGE:
[PASTE CUSTOMER MESSAGE HERE]"""

key_style_features = [
    'A machine-readable contract: the output is a strict JSON object with a fixed schema and the instruction to emit nothing outside it, so the reply can be parsed and routed by the support system automatically.',
    'Deterministic rules instead of open discussion: numbered operating rules, a closed category list, a hard word limit, an explicit grounding rule on CONTEXT, and an escalate flag that triggers on refunds, credentials or missing information.',
]

justification = """Functional fits best because this prompt is a component inside a product, not a
conversation with a human prompter. The chatbot runs unattended at scale, so its output must be
predictable, parseable and safe: the ticketing system needs a category to route on, an escalate flag
to hand off to a human, and a bounded reply to display in the widget.

Exploratory would be wrong: it is designed to open up options and brainstorm, which produces variable,
unbounded answers that no downstream system can consume. Structured is closer, since it imposes
sections and format, but it still targets a document a person reads and reviews, with no guarantee of
machine-readable output or runtime safety rules. Conversational would match the chat surface but
optimises for a natural free-flowing dialogue, which here invites hallucinated policy answers,
inconsistent tone and promises the company cannot keep.

Functional is the only style that combines a fixed I/O schema, explicit guardrails and grounding in
retrieved context, which is exactly what a customer-support chatbot in production needs."""

chosen_style, key_style_features
```

## Exercise 5: Prompt Refinement Challenge ? Control Style, Structure, and Length
**Scenario**: Write a prompt for PulseOne Mini smartwatch email blurb.

## Exercise 6: Hallucination Spotting and Mitigation
**Scenario**: Model added an unsupported claim in a climate/marine article summary.

### Tasks
1) Prompt the model to avoid hallucinations using only provided text.
2) Rewrite with verification language.
3) List 2 mitigation strategies.
4) Name 2 domains where hallucinations are risky and why.

```python
no_hallucination_prompt = """You are a science editor summarizing an article for a general audience.

Summarize the article provided below.

Grounding rules:
- Use only information explicitly stated in the article. Do not add background knowledge,
  context, causes, consequences or examples that are not in the text.
- Do not add any number, percentage, date, location, species, organisation or named person
  that does not appear verbatim in the article.
- Do not generalize: if the article describes one reef, one study or one region, do not extend
  the claim to oceans, the planet or the future.
- If the article does not answer an obvious question, write "The article does not say."
  instead of filling the gap.
- Remove any sentence you cannot trace back to a specific sentence in the article.

Format: 4 bullet points maximum, one sentence each, neutral tone.

ARTICLE:
[PASTE ARTICLE HERE]"""
no_hallucination_prompt
```

```python
verification_prompt = """You are a fact-checking editor. You will summarize the article below and then
verify your own summary against it.

Step 1 - Draft: write a summary of the article in at most 4 bullet points, using only the article text.

Step 2 - Verify: for every bullet, output a verification block in this exact format:
    CLAIM: <the bullet, shortened>
    EVIDENCE: "<exact quote copied word for word from the article>"
    STATUS: SUPPORTED | PARTIALLY SUPPORTED | UNSUPPORTED

Step 3 - Correct: delete or rewrite every claim whose status is not SUPPORTED, then output the final
summary under the heading "VERIFIED SUMMARY".

Rules:
- Every quote in EVIDENCE must be copyable from the article. If you cannot find a quote, the status
  is UNSUPPORTED and the claim must be removed.
- Never infer, extrapolate or add a figure, date, place or causal link that is not in the article.
- If the article is too short or vague to support any claim, output exactly:
  "Insufficient information in the source to summarize."
- Flag any sentence you are unsure about with [UNCERTAIN] rather than presenting it as fact.

ARTICLE:
[PASTE ARTICLE HERE]"""
verification_prompt
```

```python
mitigation_strategies = [
    'Grounding plus explicit abstention: supply the source text in the prompt, restrict the model to that text only, and give it a licensed way out ("The article does not say", "NOT IN SOURCE") so inventing is never the easiest path. In production this is retrieval-augmented generation with a citation required for each claim.',
    'Self-verification and human review: make the model quote the exact supporting sentence for every claim and label it SUPPORTED / UNSUPPORTED, then have a second pass (another model call or a human expert) check the quotes against the source before publication. A low temperature and a required confidence score help route doubtful outputs to review.',
]

risky_domains = {
    'domain_1': {
        'name': 'Healthcare and medicine',
        'why': 'An invented dosage, drug interaction, contraindication or diagnosis can directly cause physical harm, and patients or clinicians may act on it immediately with no obvious signal that the fabricated detail is false.',
    },
    'domain_2': {
        'name': 'Law and regulatory compliance',
        'why': 'Fabricated statutes, case citations or filing deadlines look authoritative but are unverifiable at a glance; relying on them can lose a case, trigger sanctions for the lawyer, or put a company in breach of regulation with financial and legal liability.',
    },
}
mitigation_strategies, risky_domains
```
