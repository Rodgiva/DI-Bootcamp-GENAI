## 🌟 Exercise 1: Basic Usage of SciPy

```python
from scipy import stats
```

## 🌟 Exercise 2: Descriptive Statistics

```python
data = [12, 15, 13, 12, 18, 20, 22, 21]
mean_data = stats.tmean(data)
median_data = stats.scoreatpercentile(data, 50)
variance_data = stats.tvar(data)
deviation_data = stats.tstd(data)

print(f"mean: {mean_data}")
print(f"median: {median_data}")
print(f"variance: {variance_data}")
print(f"deviatio: {deviation_data}")
```

## 🌟 Exercise 3: Understanding Distributions

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

mu, sigma = 50, 10 # mean and standard deviation
gaussian_dist = stats.norm(mu, sigma)
print("PDF at x=0.5:", gaussian_dist.pdf(0.5))

x_gaussian = np.linspace(mu - 50, mu + 50, 100)
pdf_gaussian = gaussian_dist.pdf(x_gaussian)
plt.figure(figsize=(6, 4))
plt.plot(x_gaussian, pdf_gaussian)
plt.title(f'Gaussian Distribution\nmu={mu}, sigma={sigma}')
plt.xlabel('x')
plt.ylabel('PDF')
plt.show()
```

## 🌟 Exercise 4: T-Test Application

```python
data1 = np.random.normal(50, 10, 100)
data2 = np.random.normal(60, 10, 100)
alpha_level = 0.5

t_statistic, p_value = stats.ttest_ind(data1, data2)
print("T-statistic:", t_statistic)
print("P-value:", p_value)

if p_value < alpha_level:
    print(f"there is a significant difference between the two groups")
```

## 🌟 Exercise 5: Linear Regression Analysis

```python
house_sizes = [50, 70, 80, 100, 120]
house_prices = [150000, 200000, 210000, 250000, 280000]

slope, intercept, r_value, p_value, std_err = stats.linregress(house_sizes, house_prices)

print("Slope:", slope)
print("Intercept:", intercept)

price = slope * 90 + intercept
print(f"Price: {price}")
```

The slope of means each additional square meter of house size is associated with an average price increase of about 1828.77 currency units.

## 🌟 Exercise 6: Understanding ANOVA

```python
fertilizer_1 = [5, 6, 7, 6, 5]
fertilizer_2 = [7, 8, 7, 9, 8]
fertilizer_3 = [4, 5, 4, 3, 4]

f_value, p_value = stats.f_oneway(fertilizer_1, fertilizer_2, fertilizer_3)
print("F-value:", f_value)
print("P-value:", p_value)

if p_value < 0.5:
    print(f"The fertilizers have significantly different effects on plant growth")
else:
    print(f"Fail to reject the null hypothesis, meaning there wouldn't be enough statistical evidence to conclude that the fertilizers have significantly different effects on plant growth.")
```
