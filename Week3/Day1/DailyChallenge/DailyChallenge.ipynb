Step 2

- Data analyzed: The team cross-referenced U.S. customs and trade records (via Import Genius, Panjiva, Volza) with on-the-ground environmental and health data — soil, dust, and air samples, plus blood tests from 70 volunteers in Ogijo, Nigeria, compared against a control group.
- Methods used: They combined epidemiological comparison (exposed population vs. control group), supply-chain tracing (following individual shipping containers from Nigerian factories through Baltimore to U.S. battery makers), and a documentary review of decades of internal industry records showing prior knowledge of the risks.
- Outcomes and decisions: The analysis showed roughly 70% of tested residents had harmful blood-lead levels, which triggered Nigeria to shut down recycling factories, launch new testing and relocation efforts, and led East Penn Manufacturing to stop sourcing lead from Nigeria.

Step 4

Without data analysis, this story would have remained a suspicion without proof. By cross-referencing customs records with blood and environmental tests conducted on site, journalists transformed a general intuition about the dangers of lead recycling into precise and conclusive evidence: they traced every shipment from the Nigerian factories to the United States and found that 70% of the residents tested had dangerously high lead levels.

It was this quantifiable and traceable evidence—not just a qualitative alert—that made denial impossible and forced concrete decisions: factory closures by the Nigerian government, approval of relocation funds by the Senate, and a halt to purchases of Nigerian lead.

Data analysis thus served as proof: it transformed a problem known but ignored for decades into a legal obligation for identified stakeholders to act.
