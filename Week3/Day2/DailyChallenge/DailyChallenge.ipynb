```python
import pandas as pd
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

df = pd.read_csv('datascience_salaries.csv')
df.head()
```

```python
normalized_df = (df['salary']-df['salary'].min())/(df['salary'].max()-df['salary'].min())

normalized_df.head(100)
```

```python
cat_cols = ['job_title', 'job_type', 'experience_level', 'location', 'salary_currency']
df_encoded = pd.get_dummies(df.drop(columns=['Unnamed: 0']), columns=cat_cols)
df_encoded.head()
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(df_encoded)
```

```python
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)
```
