## 1. Data Preparation:

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

arr = np.random.uniform(-5, 35, size=(10,12))
cities = ["New York", "London", "Paris", "Tokyo", "Sydney", "Berlin", "Toronto", "Rome", "Madrid", "Singapore"]
months = ["January", "February", "March", "April", "May", "June", "July", "August", "September", "October", "November", "December"]

np.random.seed(42)
df = pd.DataFrame(arr, index=cities, columns=months)
df
```

## 2. Data Analysis:

```python
avg_annual_temp = df.mean(axis=1)
print("Annual average temperature:")
avg_annual_temp
```

```python
print("City with the highest average temperature:")
df.idxmax()
```

```python
print("City with the lowest average temperature:")
df.idxmin()
```

## 3. Data Visualization:

```python
# plt.plot(df)
# plt.xlabel('Cities')
# plt.ylabel('Temperature C°')
# plt.show()

plt.style.use('seaborn-v0_8-whitegrid')

fig, ax = plt.subplots(figsize=(12, 6))

df.plot(ax=ax, marker='o', linewidth=1.5, colormap='tab20')

ax.set_xlabel('Cities', fontsize=12, fontweight='bold')
ax.set_ylabel('Temperature (°C)', fontsize=12, fontweight='bold')
ax.set_title('Average Monthly Temperatures by City', fontsize=14, fontweight='bold')
ax.set_xticks(range(len(cities)))
ax.set_xticklabels(cities, rotation=45, ha='right')
ax.legend(title='Month', bbox_to_anchor=(1.02, 1), loc='upper left', fontsize=9)
ax.grid(True, linestyle='--', alpha=0.6)

plt.tight_layout()
plt.show()
```
