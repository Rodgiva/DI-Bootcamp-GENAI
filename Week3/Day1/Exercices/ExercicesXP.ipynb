🌟Exercise 3: Identifying Data Types

df1 — Mental health.csv

- index → qualitative: identification of the line, we never do operations on ids
- Entity → qualitative: name of the state
- Code → qualitative
- Year → qualitative but can be quantitative if we want to calculate the difference between two dates
- Schizophrenia, Bipolar disorder, Eating disorders, Anxiety disorders, Drug use disorders, Depression and Alcohol use - disorders → quantitative: they are percentage

df2 — Time Americans Spend Sleeping.csv

- index → qualitative
- Year → qualitative
- Period → qualitative: we don't do operations on this
- Avg hrs per day sleeping → quantitative
- Standard Error → quantitative
- Type of Days → qualitative
- Age Group → qualitative
- Activity → qualitative
- Sex → qualitative

🌟 Exercise 9 : Export a dataframe to excel format and JSON format.

```python
data = {
    'Name': ['Bob', 'Henry', 'Charles', 'Aurore', 'Alicia', 'Maud'],
    'Age': [28, 34, 29, 32, 31, 35],
    'City': ['New York', 'Paris', 'Berlin', 'London', 'Tokyo', 'Moscow']
}
df = pd.DataFrame(data)
df.to_json('friends.json', orient='records', indent=4)
df.to_csv('friends.csv', index=False)
```
