```python

# This piece of code is already prefilled, run it to execute it and see the results.
# It demonstrates standard classification metrics for binary loan default prediction.

from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score, average_precision_score, confusion_matrix, classification_report

# Please replace these placeholders with your true labels and predicted probabilities.
y_true = [0,1,0,1,0,0,1,0,1,0]            # placeholder labels
y_pred_proba = [0.05,0.80,0.10,0.65,0.20,0.15,0.70,0.30,0.85,0.25]  # placeholder probabilities

# You should set a decision threshold that reflects the precision–recall trade-off for your business case.
threshold = 0.5
y_pred = [1 if p >= threshold else 0 for p in y_pred_proba]

print("Accuracy:", round(accuracy_score(y_true, y_pred), 4))
print("Precision:", round(precision_score(y_true, y_pred, zero_division=0), 4))
print("Recall:", round(recall_score(y_true, y_pred, zero_division=0), 4))
print("F1-score:", round(f1_score(y_true, y_pred, zero_division=0), 4))
print("ROC-AUC:", round(roc_auc_score(y_true, y_pred_proba), 4))
print("PR-AUC (Average Precision):", round(average_precision_score(y_true, y_pred_proba), 4))
print("\nConfusion matrix:\n", confusion_matrix(y_true, y_pred))
print("\nClassification report:\n", classification_report(y_true, y_pred, zero_division=0))
```

- Next, describe an evaluation plan in complete sentences that covers the data split, the cross-validation strategy, the metrics you will report, and how you will choose a decision threshold.
    - Data split: We will split the dataset into three parts: a training set (70%), used to fit the model, a validation set (20%), used to tune the model and compare different configurations, and a held-out test set (10%), used only once at the end to report the final, unbiased performance. If we evaluated the model on the same data it was trained on, it would simply memorize the answers instead of learning to generalize, which would give a misleadingly perfect performance.
    - Cross-Validation: to share out all the datas and try to avoid overly risky results, he training set is divided into k equal parts (folds). The model is trained and evaluated k times: each time, one fold is held out for validation, and the remaining k-1 folds are used for training. Each run starts with a fresh, untrained model, there is no memory carried over between rounds.

    | Turn (k) | Validation (100 lines) | Training (400 lines) |
    |------|--------------------------|----------------------------|
    | 1 | Fold 1 (lignes 1-100) | Folds 2,3,4,5 |
    | 2 | Fold 2 (lignes 101-200) | Folds 1,3,4,5 |
    | 3 | Fold 3 (lignes 201-300) | Folds 1,2,4,5 |
    | 4 | Fold 4 (lignes 301-400) | Folds 1,2,3,5 |
    | 5 | Fold 5 (lignes 401-500) | Folds 1,2,3,4 |
        
    - Metrics
        - Accuracy = % of correct predictions overall
        - Precision = among clients predicted as "default," how many actually default
        - Recall = among clients who actually default, how many does the model catch
        - F1-score = balanced average of precision and recall
        - ROC-AUC = measures the model's overall ability to distinguish between the two classes, regardless of the chosen threshold.
        - PR-AUC = similar but more informative than ROC-AUC when classes are imbalanced
    - Decision Threshold: the probability cutoff used to convert a model's predicted probability into a final class decision.
